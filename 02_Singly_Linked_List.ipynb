```python
class Node:
    def __init__(self, x):
        self.value = x
        self.next = None
```

```python
class LinkedList:
    def __init__(self):
        self.__head = None
        self.__tail = None
        self.__size = 0

    def insert_first(self, x):
        node = Node(x)
        node.next = self.__head
        self.__head = node
        if self.__size == 0:
            self.__tail = node
        self.__size += 1

    def len(self):
        return self.__size

    def iter_seq(self):
        node = self.__head
        while node is not None:
            yield node.value
            node = node.next

    def build(self, X):
        self.__head = None
        self.__tail = None
        self.__size = 0

        for i in range(len(X)-1, -1, -1):
            self.insert_first(X[i])

    def __get_node_at(self, i):
        if i < 0 or i >= self.__size:
            raise IndexError
        elif i == self.__size - 1:
            return self.__tail

        node = self.__head
        for _ in range(i):
            node = node.next
        
        return node

    def get_at(self, i):
        return self.__get_node_at(i).value

    def set_at(self, i, x):
        self.__get_node_at(i).value = x

    def delete_first(self):
        if self.__size == 0:
            raise IndexError

        node = self.__head
        self.__head = node.next
        self.__size -= 1
        if self.__size == 0:
            self.__tail = None

        return node.value

    def insert_last(self, x):
        if self.__size == 0:
            self.insert_first(x)
        else:
            node = Node(x)
            self.__tail.next = node
            self.__tail = node
            self.__size += 1

    def delete_last(self):
        if self.__size == 0:
            raise IndexError
        elif self.__size == 1:
            return self.delete_first()
        else:
            node = self.__get_node_at(self.__size - 2)
            last_value = node.next.value
            node.next = None
            self.__tail = node
            self.__size -= 1
            return last_value

    def insert_at(self, i, x):
        if i < 0 or i > self.__size:
            raise IndexError
        elif i == 0:
            self.insert_first(x)
        elif i == self.__size:
            self.insert_last(x)
        else:
            node = self.__get_node_at(i - 1)
            new_node = Node(x)
            new_node.next = node.next
            node.next = new_node
            self.__size += 1

    def delete_at(self, i):
        if i < 0 or i >= self.__size:
            raise IndexError
        elif i == 0:
            return self.delete_first()
        elif i == self.__size - 1:
            return self.delete_last()
        else:
            node = self.__get_node_at(i - 1)
            value = node.next.value
            node.next = node.next.next
            self.__size -= 1
            return value
```

```python
linked = LinkedList()

linked.build([10, 20, 30])
assert linked.len() == 3
assert list(linked.iter_seq()) == [10, 20, 30]

assert linked.get_at(0) == 10
assert linked.get_at(2) == 30

linked.set_at(1, 99)
assert list(linked.iter_seq()) == [10, 99, 30]

linked.insert_first(5)
linked.insert_last(40)
linked.insert_at(2, 15)

assert list(linked.iter_seq()) == [5, 10, 15, 99, 30, 40]

assert linked.delete_first() == 5
assert linked.delete_last() == 40
assert linked.delete_at(1) == 15

assert list(linked.iter_seq()) == [10, 99, 30]
assert linked.len() == 3

# Single-node and empty-list cases
linked.build([100])
assert linked.delete_last() == 100
assert linked.len() == 0

linked.insert_last(200)
assert linked.delete_at(0) == 200
assert linked.len() == 0

print("All tests passed!")
```
