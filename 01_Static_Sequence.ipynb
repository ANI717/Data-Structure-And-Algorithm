A static sequence has a fixed length after construction, but its elements can be modified.

```python
import ctypes


class StaticSequence:
    def __init__(self, size):
        if size <= 0:
            raise ValueError
        
        self.__size = size
        self.__data = (ctypes.py_object * size)()

        for i in range(size):
            self.__data[i] = None
    
    def build(self, X):
        if len(X) != self.__size:
            raise ValueError
        for i in range(self.__size):
            self.__data[i] = X[i]

    def len(self):
        return self.__size
    
    def iter_seq(self):
        for i in range(self.__size):
            yield self.__data[i]

    def get_at(self, i):
        if i < 0 or i >= self.__size:
            raise IndexError
        return self.__data[i]

    def set_at(self, i, x):
        if i < 0 or i >= self.__size:
            raise IndexError
        self.__data[i] = x
```

```python
seq = StaticSequence(5)
seq.build([10, 20, 30, 40, 50])

assert seq.len() == 5
assert seq.get_at(2) == 30

seq.set_at(2, 99)

assert seq.get_at(2) == 99
assert seq.len() == 5
assert list(seq.iter_seq()) == [10, 20, 99, 40, 50]

print("All tests passed!")
```
